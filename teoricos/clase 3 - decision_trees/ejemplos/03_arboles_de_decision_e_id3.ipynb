--- 
## 2. Entropía y Selección del Nodo Raíz

```python
features = ['Cielo', 'Temperatura', 'Humedad', 'Viento']
target = 'Jugar'

# Evaluación de ganancia de información en la raíz
print_entropy_gain_table(tennis_data, target, features)
```

--- 
## 3. Entrenamiento del Árbol ID3 y Visualización

```python
# Entrenar modelo ID3
tree = ID3DecisionTree(criterion='gain', min_info_gain=0.0).fit(tennis_data, target_attr=target, features=features)

# Visualizar estructura jerárquica en texto
print_tree_ascii(tree)
```

```python
# Visualizar gráfico del árbol
plot_decision_tree(tree, figsize=(10, 6), title="Árbol de Decisión ID3 — Dataset Jugar Tenis")
```

--- 
## 4. Reglas Lógicas (Forma Normal Disyuntiva / DNF)

```python
rules = tree.to_rules()
print(f"El árbol equivale a las siguientes {len(rules)} reglas lógicas:\n")
for r in rules:
    print(f"  • {r}")
```

## 5. Manejo de Atributos Continuos (Ejemplo Juegos de Mesa)

```python
boardgames = [
    {'Genero': 'Estrategia', 'Dificultad': 30, 'Apto': 'No'},
    {'Genero': 'Aventura',   'Dificultad': 45, 'Apto': 'Si'},
    {'Genero': 'Estrategia', 'Dificultad': 32, 'Apto': 'Si'},
    {'Genero': 'Cartas',     'Dificultad': 25, 'Apto': 'Si'},
    {'Genero': 'Cartas',     'Dificultad': 70, 'Apto': 'No'},
]

# Entrenar con atributo continuo 'Dificultad'
tree_cont = ID3DecisionTree(
    continuous_attributes={'Dificultad'},
    min_info_gain=0.01
).fit(boardgames, target_attr='Apto')

print_tree_ascii(tree_cont)
```
